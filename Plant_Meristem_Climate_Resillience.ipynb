```python
import numpy as np
import pandas as pd

def calculate_resilience_score(stress_df):
    """
    Computes a Meristem Climate Resilience and Lodging Risk Score
    based on thermal stress (heat) and waterlogging duration.
    """
    # Weights for abiotic stress factors
    heat_weight = 0.55
    water_weight = 0.45

    # Normalizing stress metrics (assuming inputs are scaled 0-100)
    thermal_stress = stress_df.get('heat_wave_intensity', 0)
    waterlogging = stress_df.get('water_retention_days', 0)

    # Calculating lodging risk index (higher score = higher risk of falling over)
    lodging_risk = (thermal_stress * heat_weight) + (waterlogging * water_weight)

    # Inverting risk to derive a Resilience & Longevity Score (0 to 100 scale)
    resilience_score = np.clip(100 - lodging_risk, 0, 100)

    stress_df['Lodging_Risk_Index'] = lodging_risk
    stress_df['Meristemic_Resilience_Score'] = resilience_score

    return stress_df

if __name__ == "__main__":
    # Sample mock data for regional crop stress tests (Banana, Sugarcane, Rice)
    sample_data = pd.DataFrame({
        'crop_type': ['Sugarcane', 'Banana', 'Rice'],
        'heat_wave_intensity': [75.0, 85.0, 60.0],
        'water_retention_days': [6.0, 9.0, 12.0]
    })

    results = calculate_resilience_score(sample_data)
    print("--- Crop Meristem Resilience Analysis ---")
    print(results[['crop_type', 'Lodging_Risk_Index', 'Meristemic_Resilience_Score']])
```
